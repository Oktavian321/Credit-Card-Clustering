## Ringkasan
Project ini merupakan credit card clustering. Dataset diperoleh dari kaggle yang terdiri atas 9000 data dan 18 variabel. pada dataset ini ada beberapa missing value pada variabel 'MINIMUM_PAYMENTS'. 4. Profiling cluster

# Data Cleaning

## Pastikan dulu minimum payment itu kosong karena apa

```python
df[df['MINIMUM_PAYMENTS'].isnull()][['BALANCE', 'PURCHASES', 'PAYMENTS']].describe()
```

## Isi nilai kosong dengan 0

```python
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(0)
```

## Isi nilai kosong pada 'CREDIT_LIMIT' dengan median

```python
df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(df['CREDIT_LIMIT'].median())
```

```python
df.isnull().sum()
```

## Drop data yang gak penting

```python
df = df.drop(columns='CUST_ID')
```

## Clip data

```python
df['CASH_ADVANCE_FREQUENCY'] = df['CASH_ADVANCE_FREQUENCY'].clip(upper=1)
```

## Log transform fitur yang skew berat.

```python
log_cols = ['MINIMUM_PAYMENTS', 'ONEOFF_PURCHASES', 'PURCHASES',
            'INSTALLMENTS_PURCHASES', 'PAYMENTS', 'CASH_ADVANCE_TRX',
            'CASH_ADVANCE', 'CASH_ADVANCE_FREQUENCY' ,'PURCHASES_TRX',
            'BALANCE', 'CREDIT_LIMIT']

df[log_cols] = np.log1p(df[log_cols])


for kolom in kolom_num:
    plt.figure(figsize=(6,5))
    sns.histplot(df[kolom], kde=True, bins=30)
    plt.title(f'Histogram {kolom}')
    plt.show()
    plt.close()
print(df[kolom_num].skew().sort_values(ascending=False))
```

## STANDARD SCALER

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

df.head().T
```

---------------------------------
## Cek elbow untuk menentukan banyaknya K

```python
model = KMeans(random_state = 42,  n_init=10)

viz = KElbowVisualizer(model, k=(2, 11), metric='distortion', timings=False)
viz.fit(X_scaled)
viz.show()

print(f'Nilai K: {viz.elbow_value_}')
```

-------------------------------
## jalankan model KMeans

```python
k = viz.elbow_value_

kmeans = KMeans(n_clusters=k, random_state = 42, n_init=10)
labels = kmeans.fit_predict(X_scaled)
```

----------------------------------------
## Evaluasi model

```python
print('Silhouette :', silhouette_score(X_scaled, labels))
print('Davies-Bouldin :', davies_bouldin_score(X_scaled, labels))
print('Calinski_Harabasz :', calinski_harabasz_score(X_scaled, labels))
```
